## 1. Liste: creazione, accesso, slicing

Una lista è una sequenza ordinata e modificabile, scritta fra parentesi quadre. Gli indici partono da zero,
come negli array Java, ma sono ammessi anche indici negativi che contano dalla fine.

```python
voti = [7, 5.5, 8, 6, 9]
aule = ["A12", "A14", "Lab. informatica"]
vuota = []

print(voti)
print(len(voti), len(aule), len(vuota))
print(voti[0], voti[2], voti[-1], voti[-2])

voti[1] = 6
print(voti)
```

```python
giorni = ["lun", "mar", "mer", "gio", "ven", "sab"]

print(giorni[1:4])   # dal secondo al quarto
print(giorni[:3])    # i primi tre
print(giorni[3:])    # dal quarto in poi
print(giorni[::2])   # uno sì e uno no
print(giorni[::-1])  # ordine invertito
print(giorni[2:2])   # slice vuoto: nessun errore
```

## 2. Modificare una lista

A differenza di un array Java, una lista cambia dimensione durante l'esecuzione. Va tenuta ferma la differenza
fra `append`, che aggiunge un solo elemento qualunque esso sia, ed `extend`, che aggiunge uno per uno gli
elementi di un'altra sequenza.

```python
presenti = ["Bianchi", "Rossi"]
presenti.append("Verdi")
presenti.insert(0, "Amato")
presenti.extend(["Zanella", "Conti"])
print(presenti)

presenti.remove("Rossi")
ultimo = presenti.pop()
print(ultimo, presenti)

a = [1, 2]
b = [1, 2]
a.append([3, 4])
b.extend([3, 4])
print(a, b)
```

## 3. Riferimenti, alias e copie

Una variabile non contiene la lista: contiene un riferimento all'oggetto lista. L'assegnamento fra variabili
copia il riferimento, quindi i due nomi indicano lo stesso oggetto. È l'errore più insidioso del modulo:
non produce alcun messaggio, solo un risultato sbagliato.

```python
a = [1, 2, 3]
b = a
b.append(4)
print("a =", a)
print("b =", b)
print("stesso oggetto?", a is b)

originale = [3, 1, 2]
copia = originale.copy()
copia.append(9)
print(originale, copia, originale is copia)
```

```python
# la copia con copy() è superficiale: gli elementi annidati restano condivisi
tabellone = [[0, 0], [0, 0]]
superficiale = tabellone.copy()
superficiale[0].append(5)
print("originale  ", tabellone)
print("superficiale", superficiale)
```

## 4. Ordinare: sort, sorted e il parametro key

Il metodo `sort` ordina sul posto e restituisce `None`; la funzione `sorted` restituisce una lista nuova e
lascia intatto l'argomento. Il parametro `key` indica secondo quale valore confrontare gli elementi. La
scrittura `lambda m: m[1]` significa «dato un elemento `m`, confronta `m[1]`»: per ora va usata solo dentro
`key`, la spiegazione completa arriva nel modulo M4.

```python
voti = [7, 5, 9, 6]
print(sorted(voti), voti)
voti.sort()
print(voti)
print("sort restituisce:", voti.sort())

parole = ["variabile", "if", "ciclo", "dizionario"]
print(sorted(parole, key=len))

studenti = ["Rossi Marco", "bianchi Ada", "Conti Luca"]
print(sorted(studenti))
print(sorted(studenti, key=str.lower))
```

```python
misure = [("aula 1", 21.5), ("aula 2", 19.0), ("aula 3", 23.2)]
print(sorted(misure, key=lambda m: m[1]))
print(sorted(misure, key=lambda m: m[1], reverse=True))

esiti = [("Rossi", 7), ("Bianchi", 9), ("Conti", 7), ("Amato", 9)]
print(sorted(esiti, key=lambda e: (-e[1], e[0])))
```

## 5. Tuple, spacchettamento, enumerate e zip

La tupla è una sequenza immutabile: si usa quando un gruppo di valori è unitario e non deve cambiare.
Lo spacchettamento assegna in un colpo solo i suoi elementi a più variabili.

```python
riga_log = ("2026-11-23", "ERRORE", "connessione rifiutata")
data, livello, messaggio = riga_log
print(f"[{livello}] {data}: {messaggio}")

uno = (7,)
non_tupla = (7)
print(type(uno), type(non_tupla))

x = 1
y = 2
x, y = y, x
print(x, y)
```

```python
nomi = ["Rossi", "Bianchi", "Conti"]
voti = [7, 5, 9]

for posizione, voto in enumerate(voti, start=1):
    print(f"{posizione}. {voto}")

for nome, voto in zip(nomi, voti):
    print(f"{nome}: {voto}")
```

## 6. Dizionari

Il dizionario associa chiavi univoche e immutabili a valori qualsiasi. L'operatore `in` lavora sulle chiavi.
Il metodo `get` restituisce un valore di riserva invece di interrompere il programma con `KeyError`, e
`setdefault` inserisce la chiave se manca: è lo strumento standard per raggruppare dati per categoria.

```python
voto = {"Rossi": 7, "Bianchi": 5, "Conti": 9}
voto["Verdi"] = 8
voto["Bianchi"] = 6
print(voto, len(voto), "Neri" in voto)

for nome, v in voto.items():
    print(f"{nome}: {v}")

print("media:", sum(voto.values()) / len(voto))
print(voto.get("Neri"), voto.get("Neri", 0), voto.get("Rossi", 0))
```

```python
registro = [("Rossi", 7), ("Bianchi", 5), ("Rossi", 9), ("Conti", 6), ("Bianchi", 8)]

voti_per_studente = {}
for nome, v in registro:
    voti_per_studente.setdefault(nome, []).append(v)
print(voti_per_studente)

for nome, elenco in voti_per_studente.items():
    media = sum(elenco) / len(elenco)
    print(f"{nome}: {len(elenco)} voti, media {media:.2f}")
```

## 7. Insiemi

L'insieme conserva elementi distinti, non ha indici e non garantisce l'ordine di visualizzazione: se serve un
ordine si passa da `sorted`. L'insieme vuoto si scrive `set()`, perché `{}` crea un dizionario vuoto.

```python
linguaggi_4a = {"Java", "Python", "SQL"}
linguaggi_4b = {"Python", "C", "SQL", "Python"}
print(len(linguaggi_4b), "Java" in linguaggi_4a)

print("unione      ", sorted(linguaggi_4a | linguaggi_4b))
print("intersezione", sorted(linguaggi_4a & linguaggi_4b))
print("differenza  ", sorted(linguaggi_4a - linguaggi_4b))
print("solo in uno ", sorted(linguaggi_4a ^ linguaggi_4b))
```

## 8. Comprehension

La comprehension costruisce una nuova collezione combinando in una riga il ciclo, la trasformazione e il
filtro. Con le parentesi quadre produce una lista, con le graffe un dizionario oppure un insieme.

```python
voti = [7, 5, 9, 6, 8]
print([v for v in voti if v >= 6])
print([v * 10 for v in voti])

nomi = ["Rossi", "Bianchi", "Conti"]
registro = {n: v for n, v in zip(nomi, [7, 5, 9])}
print(registro)
print({n: len(n) for n in nomi})
print(sorted({n[0] for n in nomi}))
print({n: v for n, v in registro.items() if v >= 6})
```

## 9. Ricerca, ordinamento e costo delle operazioni

Le due ricerche note da Java si riscrivono in Python contando i confronti effettivamente eseguiti. La ricerca
lineare funziona su qualunque lista e costa O(n); la ricerca binaria pretende dati ordinati e costa O(log n),
perché ogni passo dimezza l'intervallo da esaminare.

```python
matricole = [1042, 1077, 1103, 1128, 1150, 1181, 1204]
cercata = 1150

posizione = -1
confronti = 0
for i in range(len(matricole)):
    confronti += 1
    if matricole[i] == cercata:
        posizione = i
        break
print(f"lineare: posizione {posizione}, confronti {confronti}")
```

```python
sinistra = 0
destra = len(matricole) - 1
posizione = -1
confronti = 0
while sinistra <= destra:
    medio = (sinistra + destra) // 2
    confronti += 1
    if matricole[medio] == cercata:
        posizione = medio
        break
    elif matricole[medio] < cercata:
        sinistra = medio + 1
    else:
        destra = medio - 1
print(f"binaria: posizione {posizione}, confronti {confronti}")
```

```python
# ordinamento per selezione, con il conteggio dei confronti
voti = [7, 5, 9, 6, 8]
confronti = 0
for i in range(len(voti)):
    minimo = i
    for j in range(i + 1, len(voti)):
        confronti += 1
        if voti[j] < voti[minimo]:
            minimo = j
    voti[i], voti[minimo] = voti[minimo], voti[i]
print(voti, confronti)
```
