### Esercizio 2

```python
studente = "Gianni"
voti = [8.5, 7.25, 8.75]
assenza = 45
somma = 0

for voto in voti:
    somma += voto

media = somma / len(voti)

if media >= 8:
    risultato = "ottimo"
elif media >= 6:
    risultato = "sufficiente"
elif media < 6:
    risultato = "insufficiente"

print(f"{studente}: media {media:.2f}, giudizio {risultato}")
```

### Esercizio 3

```python
inizio = 15
fine = 32
passo = 2

# 2. Controllo valore variabile passo
if passo <= 0:
    print("Errore: il passo deve essere un valore positivo.")
else:
    # Definizione delle larghezze delle colonne per un allineamento perfetto
    w_km = 10
    w_miglia = 12
    w_metri = 10

    # 3. Stampa dell'intestazione della tabella (allineata a destra)
    print(f"{'km':>{w_km}} {'miglia':>{w_miglia}} {'metri':>{w_metri}}")
    
    # Stampa della riga di trattini (calcolata in base alla larghezza totale)
    lunghezza_totale = w_km + w_miglia + w_metri + 2  # +2 per gli spazi intermedi
    print("-" * lunghezza_totale)

    # 4. Ciclo per calcolare e stampare i valori dell'intervallo (estremi inclusi)
    # Usiamo fine + 1 nel range per includere l'estremo superiore
    for km in range(inizio, fine + 1, passo):
        # Calcolo delle conversioni
        miglia = km * 0.621371
        metri = km * 1000
        
        # Stampa della riga formattata:
        # :>{w_km} allinea a destra con la larghezza scelta
        # :>{w_miglia}.3f allinea a destra e impone esattamente 3 decimali
        print(f"{km:>{w_km}} {miglia:>{w_miglia}.3f} {metri:>{w_metri}}")
```

### Esercizio 4

```python
# Lettura dell'input da tastiera
ingresso = "LAB2-PC17-2026"

# Ripulitura dagli spazi agli estremi e conversione in maiuscolo
codice = ingresso.strip().upper()

# Estrazione delle tre parti tramite slicing
laboratorio = codice[0:4]
calcolatore = codice[5:9]
anno = codice[10:14]

# Stampa delle tre parti su righe distinte
print(f"laboratorio: {laboratorio}")
print(f"calcolatore: {calcolatore}")
print(f"anno: {anno}")

# Calcolo della lunghezza e dell'ultimo carattere con indice negativo
lunghezza = len(codice)
ultimo_carattere = codice[-1]

# Stampa finale della lunghezza e dell'ultimo carattere
print(f"lunghezza: {lunghezza}, ultimo carattere: {ultimo_carattere}")
```

### Esercizio 5

```python
# Lettura della frase in ingresso
frase = "Il laboratorio 4BI dura 22 ore"

# Inizializzazione dei contatori
vocali = 0
consonanti = 0
cifre = 0
altri = 0

# Definizione delle vocali (sia minuscole che maiuscole)
stringa_vocali = "aeiouAEIOU"

# Unico ciclo for sui caratteri della frase
for carattere in frase:
    if carattere in stringa_vocali:
        vocali += 1
    elif carattere.isalpha():
        # NOTA RICHIESTA: Grazie a .isalpha(), le lettere accentate (es. à, è, é) 
        # non incluse nella stringa delle vocali vengono considerate come consonanti.
        consonanti += 1
    elif carattere.isdigit():
        cifre += 1
    else:
        altri += 1

# Calcolo del totale dei caratteri e della percentuale di vocali
totale_caratteri = vocali + consonanti + cifre + altri

# Gestione del caso in cui la frase sia vuota per evitare la divisione per zero
if totale_caratteri > 0:
    percentuale_vocali = (vocali / totale_caratteri) * 100
else:
    percentuale_vocali = 0.0

# Stampa dei risultati secondo il formato richiesto
print(f"vocali: {vocali}, consonanti: {consonanti}, cifre: {cifre}, altri: {altri}")
print(f"percentuale di vocali: {percentuale_vocali:.1f}%")
```

### Esercizio 7

```python
# Lettura del numero intero non negativo
numero = int(input("Inserisci un numero intero non negativo: "))

# Inizializzazione dei contatori
somma_cifre = 0
conteggio_cifre = 0
cifra_maggiore = 0

# Gestione specifica per il caso in cui il numero sia esattamente 0
if numero == 0:
    conteggio_cifre = 1
    somma_cifre = 0
    cifra_maggiore = 0
else:
    # Creiamo una copia del numero per non modificare la variabile originale
    n = numero
    
    # Ciclo while per analizzare il numero cifra per cifra da destra a sinistra
    while n > 0:
        cifra = n % 10            # Estrae l'ultima cifra a destra (resto della divisione per 10)
        
        somma_cifre += cifra      # Aggiunge la cifra alla somma totale
        conteggio_cifre += 1      # Incrementa il numero di cifre trovate
        
        # Controlla se la cifra corrente è la più grande trovata finora
        if cifra > cifra_maggiore:
            cifra_maggiore = cifra
            
        n = n // 10               # Rimuove l'ultima cifra a destra (divisione intera per 10)

# Stampa dei risultati secondo il formato richiesto
print(f"cifre: {conteggio_cifre}")
print(f"somma delle cifre: {somma_cifre}")
print(f"cifra maggiore: {cifra_maggiore}")
```
